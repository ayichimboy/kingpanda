```python
import numpy as np 
import pandas as pd       
import matplotlib.pyplot as plt        
import seaborn as sns     

```

```python
df = pd.read_csv("sample_drone_flights.csv")
```

```python
df["wind_direction(degrees)"] = df["wind_direction(degrees)"].fillna(df["isPhoto"])
```

```python
df["message"] = df["message"].fillna(df["isPhoto"])
```

"""
apply takes in series, one column at a time .
when you use apply, you're telling the function, while iterating over 
each individual column, do something to the columns

so when a function uses apply, what you're saying is that do something 
to indivivual columns instead of the whole dataframe.



"""



```python
def det(df):
    df.columns =  df.columns.str.lower()
    return df
df["flycState"].apply(lambda x: x.lower())
```

```python
df["battery_temp_degree_c"] = df["battery_temperature(f)"].apply(lambda x: (x-30)/2 )
```

```python
df[["battery_temp_degree_c", "battery_temperature(f)"]]
```

```python
df["datetime(utc)"] = pd.to_datetime(df['datetime(utc)'])
calendar = ["day", "month", "year"]

for feature in calendar:
    df[feature] = getattr(df["datetime(utc)"].dt, feature)
    
# df
```
